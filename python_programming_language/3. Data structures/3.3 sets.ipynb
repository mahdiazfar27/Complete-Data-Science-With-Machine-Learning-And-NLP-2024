### Sets

## like c++, pythonic set is not sorted, rather it stores elements in a arbitary order
in sets items are unordered, unique element.
meaning they do not stores as inserted and no duplicate element.
this is useful for union,intersection,difference,symmetric difference

```python
### create a set
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
```

```python
### typcasting to set
my_set = set([1,2,3,4,5]) # convert list to set
print(my_set)
print(type(my_set))
```

```python
my_set = set([5,6,7,5,6,7,6,5,6,7,7,6]) # convert list to set and remove duplicates
print(my_set)
```

```python
## basic set operation are union, intersection, difference and symmetric difference
# union is the set of all elements that are in either set A or set B or both. it is denoted by A | B
# intersection is the set of all elements that are in both set A and set B. it is denoted by A & B
# difference is the set of all elements that are in set A but not in set B. it is denoted by A - B
# symmetric difference is the set of all elements that are in either set A or set B but not in both. it is denoted by A ^ B . its like the idea of exclusive or in logic. 

# adding element

my_set.add(8)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
# removing element
my_set.remove(7)
print(my_set)
```

```python
my_set.discard(10) # this will not raise an error because 10 is not in the set. the discard method will not raise an error if the element is not in the set. it will simply do nothing.
print(my_set)
```

```python
my_set.discard(10)
print(my_set)
```

```python
## pop arbitarily removed any element from the set and return it. since sets are unordered, we don't know which element will be removed.
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
my_set.clear()
print(my_set)
```

```python
## set membership test is used to check if an element is in a set or not. it is denoted by the keyword 'in'.
my_set = set([5,6,7,5,6,7,6,5,6,7,7,6])
print(5 in my_set)
print(10 in my_set)
```

```python
## mathematical operation

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

#union
union_set = set1.union(set2)
print(union_set)

# Intersection
inter_set = set1.intersection(set2)
print(inter_set)

# update the first set with the set operation intersection between set1 and set2 and store the result in set1.
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# Difference
diff_set = set1.difference(set2)
print(diff_set)
print(set2.difference(set1))
```

```python
## symmetric difference
print(set1.symmetric_difference(set2))
```

```python
## Set methods
set1 = {1,2,3}
set2 = {3,4,5}

#is subset
print(set1.issubset(set2))
#is superset
print(set1.issuperset(set2))
```

```python
## Set methods
set1 = {1,2,3,4,5}
set2 = {3,4,5}

#is subset
print(set1.issubset(set2))
print(set2.issubset(set1))
#is superset
print(set1.issuperset(set2))
```

```python
lst = [1,2,2,3,4,4,5]
print(set(lst))
```

```python
### counting unique words in text
text  = "in this tutorial we are discussing about sets"
words = text.split() # split the text into words and store it in a list. the default separator is any whitespace character (space, tab, newline, etc.) and it will split the string into words based on that.

unique_words = set(words)

print(unique_words)
print(len(unique_words))
```
