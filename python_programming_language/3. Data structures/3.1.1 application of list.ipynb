#### ex2 organizing student grades

```python
# write a code to organize student grades using lists. create a list of student names and a corresponding list of their grades. then, calculate the average grade and print out each student's name along with their grade and whether they passed or failed based on a passing grade of 60.
student_names = ["Alice", "Bob", "Charlie", "David", "Eve"]
student_grades = [85, 72, 90, 55, 68]
passing_grade = 60
total_grades = 0
for grade in student_grades:
    total_grades += grade
average_grade = total_grades / len(student_grades)
print(f"Average Grade: {average_grade}")
for i in range(len(student_names)):
    name = student_names[i]
    grade = student_grades[i]
    if grade >= passing_grade:
        status = "Passed"
    else:
        status = "Failed"
    print(f"{name}: {grade} - {status}")
```

#### ex3 manage inventory items in a store

```python
# write a code to manage inventory items in a store. create a list of item names and a corresponding list of their quantities. then, write a function to add new items to the inventory, update the quantity of existing items, and check if an item is in stock.
item_names = ["apple", "banana", "orange", "grape"]
item_quantities = [10, 20, 15, 5]

def add_item(name, quantity):
    item_names.append(name)
    item_quantities.append(quantity)

def update_quantity(name, new_quantity):
    if name in item_names:
        index = item_names.index(name)
        item_quantities[index] = new_quantity

def is_in_stock(name):
    if name in item_names:
        index = item_names.index(name)
        return item_quantities[index] > 0
    return False
# Example usage
add_item("pear", 12)
print(f"Inventory: {list(zip(item_names, item_quantities))}")
update_quantity("apple", 15)
print(is_in_stock("banana"))  # Output: True
print(is_in_stock("pear"))    # Output: True
```
