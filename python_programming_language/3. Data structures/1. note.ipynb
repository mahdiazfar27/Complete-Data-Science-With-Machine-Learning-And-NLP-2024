Important Python data structures

```python

from numpy.matlib import empty


list[] → ordered, mutable #initialize with square brackets: []

tuple() → ordered, immutable #initialize with parentheses: ()

set{} → unordered, unique items, mutable #initialize with braces: {}
note: empty set is created with set(), not {}


dict{} → key-value pairs, ordered(python 3.7+) but not sorted with key, mutable #initialize with braces and colons: {"key": value}
```

Python data structure initialization syntax


```python
list → square brackets

my_list = [1, 2, 3]
empty: my_list = []

tuple → parentheses
my_tuple = (1, 2, 3)
single element: my_tuple = (1,)
empty: my_tuple = ()

set → braces
my_set = {1, 2, 3}
empty: my_set = set() (not {})

dict → braces with key: value
my_dict = {"a": 1, "b": 2}
empty: my_dict = {}
        my_dict = dict()  #both are valid ways to create an empty dictionary
```

```python
C++ set ≈ Python tuple (elements immutable) 
C++ vector ≈ Python list (elements mutable) 
C++ map<Key, Value> ≈ Python dict (key-value pairs)
```
