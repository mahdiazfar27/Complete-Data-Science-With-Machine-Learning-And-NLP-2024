```python
### array module from python standard library
import array
lst = [1,2,3,4,5]
arr = array.array('i', lst) # here 'i' stands for integer the data types of the array elemtents. 
print(arr)

# array is like a list but- 
    # 1) it stores same data types only
    # 2) more memory efficient
```

```python
### json 
# this module is used for data serialization and deserialization. it is used to convert python objects to json format and json format to python objects. json stands for JavaScript Object Notation and it is a lightweight data interchange format that is easy for humans to read and write and easy for machines to parse and generate. when we retrive any data from server using API we get the data in terms of json file (no sql)

import json
data = {'name': 'farabe', 'age': 25}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

# after api call we will get this json_str
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
### csv is a module that provides functions for reading and writing csv files. csv stands for comma separated values and it is a common format for storing tabular data. we can use the csv module to read and write csv files in python. some of the commonly used functions in the csv module are reader(), writer(), DictReader(), DictWriter(), etc.
import csv
with open('example.csv', mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['farabe', 25])

with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
### datetime 
# it is a module that provides classes for manipulating dates and times. it allows us to work with date and time objects in both simple and complex ways. we can use the datetime module to perform various operations such as formatting, arithmetic, and comparison of date and time objects. some of the commonly used classes in the datetime module are datetime, date, time, timedelta, etc.

from datetime import datetime, timedelta
now = datetime.now()
print(now)
yesterday = now - timedelta(days = 2)
print(yesterday)
```

```python
### time 
# it is a module that provides various time-related functions. it allows us to work with time in both simple and complex ways. we can use the time module to perform various operations such as getting the current time, measuring the execution time of a code block, and creating delays in our code. some of the commonly used functions in the time module are time(), sleep(), ctime(), etc. 
import time
print(time.time())
time.sleep(2)
print(time.time())
```
