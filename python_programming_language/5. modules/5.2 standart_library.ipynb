```python
### json 
# this module is used for data serialization and deserialization. 
# it is used to convert python objects to json format and json format to python objects. 
# json stands for JavaScript Object Notation and it is a lightweight data interchange format that is easy for humans to read and write and easy for machines to parse and generate. 
# when we retrive any data from server using API we get the data in terms of json file (no sql)

import json
data = {'name': 'farabe', 'age': 25}

json_str = json.dumps(data) # this will convert the python dictionary data to json string format.
print(json_str)
print(type(json_str))

# after api call we will get this json_str
parsed_data = json.loads(json_str) # this will convert the json string back to python dictionary format. and we can access the data using the keys of the dictionary.
print(parsed_data)
print(type(parsed_data))
```

```python
### csv is a module that provides functions for reading and writing csv files. 
# csv stands for comma separated values and it is a common format for storing tabular data. 
# we can use the csv module to read and write csv files in python. 
# some of the commonly used functions in the csv module are reader(), writer(), DictReader(), DictWriter(), etc.
import csv
with open('example.csv', mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['farabe', 25])
    writer.writerow(['pritom', 31])

with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
### datetime 
# it is a module that provides classes for manipulating dates and times. 
# it allows us to work with date and time objects in both simple and complex ways. 
# we can use the datetime module to perform various operations such as formatting, arithmetic, and comparison of date and time objects. 
# some of the commonly used classes in the datetime module are datetime, date, time, timedelta, etc.

from datetime import datetime, timedelta
now = datetime.now()
print(now)
the_day_before_yesterday = now - timedelta(days = 2) # timedelta is a class that represents the difference between two dates or times. it can be used to perform arithmetic operations on date and time objects. here we are subtracting 2 days from the current date and time to get the date and time of the day before yesterday.
print(the_day_before_yesterday)
```

```python
### time 
# it is a module that provides various time-related functions. 
# it allows us to work with time in both simple and complex ways. 
# we can use the time module to perform various operations such as getting the current time, measuring the execution time of a code block, and creating delays in our code. 
# some of the commonly used functions in the time module are time(), sleep(), ctime() it returns the current time in seconds since the epoch etc. 
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
### re
# regular expression 
import re
pattern = r'\d+' # this pattern will match one or more digits in a string. \d stands for digit and + stands for one or more occurrences of the preceding character. so this pattern will match any sequence of digits in a string.
text = 'there are 1234 apples 456'
match = re.search(pattern, text)
print(match.group())
```
