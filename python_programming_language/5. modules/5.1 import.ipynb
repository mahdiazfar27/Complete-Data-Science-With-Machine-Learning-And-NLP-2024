### importing modules and packages in python

python hierarchy:
Library     ### collection of packages e.g. scikit-learn,pandas ### we install libraries using <pip install library_name>
    package     ### it's folders of modules e.g. sklear.preprocessing ###
        module      ### it's a single .py files e.g. math.py ###
            function/class      ### defined inside module e.g. sqrt() ### we import these packages, modules, fucn to our code using <import package/module_name as alias>
                statements      ### actual lines of codes ###

```python
### importing package 
# import package_name as alias_name 
# (alias_name is the optional name we want to use to refer to the package in our code)
import math
math.sqrt(2)
```

```python
### importing function from package
# from package_name import function_name 
# and here alias_name is not optional. we have to use the function name to refer to the function in our code.
from math import sqrt, pi
print(sqrt(25))
print(pi)
```

### here in the code:
library -> python standard library
module -> math
function -> sqrt()

```python
from math import *     
# * imports all functions from math import sqrt, pi  # this will override the previous import and we can use sqrt and pi without the math. prefix but it is not recommended to use * because it can cause conflicts with other functions that have the same name in other packages. it can also make it harder to read the code and understand where the functions are coming from. so it is better to import only the functions that we need or to use an alias for the package.

print(sqrt(36))
print(pi)
```

```python
### now importing my own module(.py file) form my own_package(folder)
# but this import.ipynb(name of this file) and the own_pakage folder shoud be in same directory

from own_package.maths import add
print(add(3,4))
```

```python
# another was to call the function from the package is to import the whole module and then call the function using the module name as a prefix. this is useful when we have multiple functions in the module and we want to avoid conflicts with other functions that have the same name in other packages.

from own_package import maths
print(maths.add(4,6))
```
