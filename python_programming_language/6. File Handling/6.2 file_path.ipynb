### file paths

```python
import os
print(os.getcwd())
```

```python
## creat a new directory
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory '{new_directory}' created")
```

```python
## listing all the files in a directory
items = os.listdir('.')
print(items)
```

```python
### joining path is a common operation when working with files and directories. it allows us to create a complete path by joining different parts of the path together. this is especially useful when we want to create a path that is platform-independent, meaning it will work on different operating systems without any issues. the os.path.join() function is used for this purpose. it takes multiple arguments and joins them together using the appropriate separator for the operating system. this way, we can ensure that our code will work correctly regardless of the platform it is run on.
# most importat feature of using os package.
dir_name = 'folder'
file_name = 'file.txt'
#relative path is the path to a file or directory relative to the current working directory. it does not include the full path from the root directory, but rather just the path from the current working directory to the file or directory. we can create a relative path by simply joining the directory name and file name using the os.path.join() function. this way, we can work with files and directories without worrying about the current working directory or the platform we are running on.
relative_path = os.path.join(dir_name,file_name)
print(relative_path)
```

```python
dir_name = 'folder'
file_name = 'file.txt'
# full path is the complete path to a file or directory, including all the directories leading up to it. we can create a full path by joining the current working directory with the relative path using the os.path.join() function. this way, we can ensure that our code will work correctly regardless of the platform it is run on.
full_path = os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
### if a file is present or not
path = "example.txt"
if os.path.exists(path):
    print(f"the path '{path}' exists")
else:
    print(f"the path '{path}' does not exist")
```

```python
### checking if a path is a file of directory
import os

path = 'example.txt'
if os.path.isfile(path):
    print(f"the path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"the path '{path}' is a directory.")
else:
    print(f"the path '{path}' is neither a file nor a directory.")
```

```python
### instead of using the join function we can also get the absolute path of a file or directory using the os.path.abspath() function. 
# this function takes a relative path as an argument and returns the absolute path of that file or directory. 
# this is useful when we want to work with files and directories without worrying about the current working directory or the platform we are running on.

relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
