Data Visualization With Matplotlib

Matplotlib is a powerful plotting library for Python that enables the creation of static, animated, and interactive visualizations. It is widely used for data visualization in data science and analytics. In this lesson, we will cover the basics of Matplotlib, including creating various types of plots and customizing them.

```python
# pyplotlib is a plotting library for Python that provides a wide range of functions for creating static, animated, and interactive visualizations. It is built on top of the NumPy library and is widely used in data analysis, scientific computing, and machine learning projects to create various types of plots and charts, such as line plots, scatter plots, bar charts, histograms, and more. Matplotlib allows you to customize the appearance of your plots and provides a high level of control over the layout and design of your visualizations.
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [1,4,9,16,25,36]
```

```python
### Create a line plot
plt.plot(x, y)
```

```python
### Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Line Plot')
plt.show() # Display the plot
```

```python
x = [1,2,3,4,5,6]
y = [1,4,9,16,25,36]

### create a customized line plot
plt.plot(x, y, marker='o', linestyle='--', color='r', linewidth=2, markersize=8)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Customized Line Plot')
plt.grid() # Add grid lines
plt.show()
```

```python
### multiple lines in one plot/one figure

x = [1,2,3,4,5,6]
y1 = [1,4,9,16,25,36]
y2 = [1,8,27,64,125,216]
y3 = [1,2,3,4,5,6]

plt.plot(x, y1, marker='o', linestyle='--', color='r', linewidth=2, markersize=8, label='y = x^2')
plt.plot(x, y2, marker='s', linestyle='-', color='b', linewidth=2, markersize=8, label='y = x^3')
plt.plot(x, y3, marker='^', linestyle='-.', color='g', linewidth=2, markersize=8, label='y = x')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Lines in One Plot')

plt.legend() # Add legend to differentiate the lines

plt.grid()

plt.show()
```

```python
# multiple plots in one figure

x = [1,2,3,4,5,6]
y1 = [1,4,9,16,25,36]
y2 = [1,8,27,64,125,216]

plt.figure(figsize=(12, 6)) # Set the figure size where 12 is the width and 6 is the height in inches.

plt.subplot(2, 2, 1) # Create the first subplot (2 rows, 2 columns, 1st plot)
plt.plot(x, y1, marker='o', linestyle='--', color='r', linewidth=2, markersize=8)
plt.title('Plot 1')

plt.subplot(2, 2, 2) # Create the second subplot (2 rows, 2 columns, 2nd plot)
plt.plot(x, y2, marker='s', linestyle='-', color='b', linewidth=2, markersize=8)
plt.title('Plot 2')

plt.subplot(2, 2, 3) # Create the third subplot (2 rows, 2 columns, 3rd plot)
plt.plot(x, y1, marker='o', linestyle='--', color='r', linewidth=2, markersize=8)
plt.plot(x, y2, marker='s', linestyle='-', color='b', linewidth=2, markersize=8)
plt.title('Plot 3')

plt.subplot(2, 2, 4) # Create the fourth subplot (2 rows, 2 columns, 4th plot)
plt.plot(x, y2, marker='o', linestyle='--', color='b', linewidth=2, markersize=8)
plt.plot(y1, y2, marker='s', linestyle='-', color='b', linewidth=2, markersize=8)
plt.title('Plot 4')

plt.tight_layout() # Adjust the spacing between subplots
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [1,4,9,16,25,36]

# Create a scatter plot
plt.scatter(x, y, color='purple', marker='x', s=100) # s is the size of the markers
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Scatter Plot')
plt.show()
```

```python
# create a bar chart plot

x = ['A', 'B', 'C', 'D', 'E', 'F']
y = [1,4,9,16,25,36]

plt.bar(x, y, color='cyan', edgecolor='black')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Bar Chart')
plt.show()
```

Histograms

Histograms are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin.

StaQuest(YT)
Here is a summary of the key points:

* **Why use histograms?** When you have too many data points to visualize individually (as they may overlap or hide each other), histograms help by dividing the data range into **bins** and stacking the values that fall within each bin (0:46–1:20).
* **Predictive Value:** Histograms allow you to visualize the probability of future measurements. Higher stacks within a bin indicate a higher frequency of occurrence, helping you decide whether to use models like a *normal distribution* or an *exponential distribution* to approximate your data (1:34–2:20).
* **The Challenge of Bin Widths:** Choosing the right bin width is critical (2:25):
    * **Bins too narrow:** Every measurement might get its own bin, providing little insight (2:29–2:43).
    * **Bins too wide:** The data might be split too crudely (e.g., 50/50), hiding the underlying patterns (2:46–3:01).
* **Takeaway:** Do not rely solely on default settings in software. Always experiment with different bin widths to find the most accurate and clear representation of your data (3:03–3:22).

```python
# Create a Pie chart

labels = ['A', 'B', 'C', 'D', 'E', 'F']
sizes_or_data = [10, 20, 30, 15, 25, 5] # Sample sizes for each category
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue', 'cyan', 'magenta'] # Optional: specify colors for each slice
explode = (0.2, 0, 0, 0, 0, 0) # Optional: "explode" or move out the first slice

plt.pie(sizes_or_data, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, explode=explode, shadow=True)

plt.title('Basic Pie Chart')
plt.show()
```

```python
# Creat a box plot
y = [1,4,9,16,25,36]
plt.boxplot(y, vert=True, patch_artist=True) # vert=True for vertical box plot, patch_artist=True to fill the box with color
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Box Plot')
plt.show()
```

```python
# create a violin plot
import matplotlib.pyplot as plt
import numpy as np
data = np.random.rand(100) # Generate a 1D array of 100 random values between 0 and 1
plt.violinplot(data, vert=True, showmeans=True) # vert=True for vertical violin plot, showmeans=True to display the mean value
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Violin Plot')
plt.show()
```

```python
# create a heatmap
import numpy as np
data = np.random.rand(10, 10) # Generate a 10x10 matrix of random values between 0 and 1
print(data) # Print the data to see the values in the matrix
plt.imshow(data, cmap='viridis', interpolation='nearest') # Display the data as an image, with the 'viridis' colormap and no interpolation
plt.colorbar() # Add a colorbar to show the scale of values
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Heatmap')
plt.show()
```

```python
# sales_data visualization
import pandas as pd

sales_data_df = pd.read_csv('100_Sales_Records.csv') # Load sales data from a CSV file

sales_data_df.head() # Display the first few rows of the sales data
```

```python
sales_data_df.info()
```

```python
# Plot total sales by Item Type
total_sales_by_item_type = sales_data_df.groupby('Item Type')['Units Sold'].sum()
print(total_sales_by_item_type)
```

```python
total_sales_by_item_type.plot(kind='bar', color='orange', edgecolor='black')
plt.xlabel('Item Type')
plt.ylabel('Total Units Sold')
plt.title('Total Units Sold by Item Type')
plt.xticks(rotation=45) # Rotate x-axis labels for better readability
plt.show()
```

```python
# Plot sales trend over time
sales_data_df['Order Date'] = pd.to_datetime(sales_data_df['Order Date'])
sales_data_df.set_index('Order Date')['Units Sold'].resample('ME').sum().plot(kind='line', color='blue', marker='o')
plt.xlabel('Date')
plt.ylabel('Total Units Sold')
plt.title('Sales Trend Over Time')
plt.show()
```

```python
# Plot sales trends over time
sales_trend = sales_data_df.groupby('Order Date')['Units Sold'].sum().reset_index() # reset_index() is used to reset the index of the DataFrame after grouping by 'Order Date' and summing 'Units Sold'. This allows us to have a clean DataFrame with 'Order Date' as a regular column instead of an index, which is useful for plotting and further analysis.
sales_trend
```

```python
sales_trend.plot(kind='line', x='Order Date', y='Units Sold', color='blue', marker='o')


plt.xlabel('Date')
plt.ylabel('Total Units Sold')
plt.title('Sales Trend Over Time')
plt.show()
```

```python
# or we can also use this syntax to plot sales trends over time
plt.plot(sales_trend['Order Date'], sales_trend['Units Sold'], marker='o', color='blue')
```
