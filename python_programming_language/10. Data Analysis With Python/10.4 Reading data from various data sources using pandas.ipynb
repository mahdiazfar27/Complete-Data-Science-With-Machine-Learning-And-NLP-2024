```python
# data read from json string 
import pandas as pd
from io import StringIO # to read the string as a file object

Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title1":"Team Lead", "title2":"Sr. Developer"}]}'
df = pd.read_json(StringIO(Data))
```

```python
# Reading data from a URL using pandas
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data" # This URL contains the Wine dataset from the UCI Machine Learning Repository which is a comma separated file with no header row.
df = pd.read_csv(url, header=None)
```

```python
df.to_csv("wine.csv", index=False) # index=False will prevent pandas from writing row indices to the CSV file.
```

```python
# Reading all the tables from a webpage using match parameter to filter tables based on a keyword in the table

url = "https://en.wikipedia.org/wiki/Mobile_country_code"
# Fetches all tables found on the Wikipedia page
df = pd.read_html(url, match="country") # match="country" will filter the tables and return only those tables which have the keyword "country" in their content. This is useful when there are multiple tables on the page and you want to extract a specific one based on its content.
```

```python
# because Wikipedia blocks the default User-Agent used by Python's scraping libraries (like urllib or requests, which pandas uses under the hood) to prevent automated scraping bot spam, returning a 403 Forbidden status code. To bypass this, we can set a custom User-Agent header that mimics a regular web browser when making the request to fetch the page content before passing it to pandas.read_html().

import pandas as pd
import requests
from io import StringIO # StringIO is used to treat the string response from requests as a file-like object that pandas can read from.

url = "https://en.wikipedia.org/wiki/Mobile_country_code"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}

response = requests.get(url, headers=headers)

# Wrap the text response inside StringIO()
df = pd.read_html(StringIO(response.text), match="country")

# Look at the first matched table
df[0]
# the match parameter is used to filter the tables based on the presence of the keyword "Country" in the table. 
# unless it will show the first table on the page.
```
