Data Visualization With Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Plot sales trend over time using Seaborn

sales_data_df = pd.read_csv('100_Sales_Records.csv')

print(sales_data_df['Order Date'])
sales_data_df['Order Date'] = pd.to_datetime(sales_data_df['Order Date']) # Convert 'Order Date' to datetime format for better handling of date-related operations
print(sales_data_df['Order Date'])

sales_trend = sales_data_df.groupby('Order Date')['Units Sold'].sum().reset_index()

sns.lineplot(data=sales_trend, x='Order Date', y='Units Sold', marker='o', color='blue')

plt.xlabel('Date')
plt.ylabel('Units Sold')
plt.title('Sales Trend Over Time')
plt.show()
```

```python
### Basic Plotting with Seaborn

tips_df = sns.load_dataset('tips') # Load the 'tips' dataset, which is a built-in dataset in Seaborn that contains information about restaurant tips, including total bill, tip amount, and other related variables. 
print(tips_df)
```

```python
# Create a scatter plot
sns.scatterplot(data=tips_df, x='total_bill', y='tip', hue='day', style='time') # hue='day' colors the points based on the day of the week, and style='time' differentiates points based on whether the meal was lunch or dinner.

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Scatter Plot of Total Bill vs Tip')

plt.legend(title='Day of Week')
plt.show()
```

```python
# Create a line plot
sns.lineplot(data=tips_df, x='size', y='total_bill', hue='day', marker='o') # hue='day' colors the lines based on the day of the week, and marker='o' adds markers to the data points for better visibility.
plt.title('Line Plot of Total Bill by Party Size')
plt.xlabel('Party Size')
plt.show()
```

```python
### Create a Categorical Bar-Plot
# Bar plot to show total bill by day of the week
sns.barplot(data=tips_df, x='day', y='total_bill')
plt.title('Average Total Bill by Day of the Week')
plt.show()
```

```python
# Box plot to show distribution of total bill by day of the week
# box plot
sns.boxplot(data=tips_df, x='day', y='total_bill')
plt.title('Distribution of Total Bill by Day of the Week')
plt.show()
```

```python
# create a violin plot
import seaborn as sns
sns.violinplot(data=tips_df, x='day', y='total_bill')
plt.title('Violin Plot of Total Bill by Day of the Week')
plt.show()
```

```python
# Create a histogram of total bill
sns.histplot(data=tips_df, x='total_bill', bins=20, kde=True) # bins=20 specifies the number of bins for the histogram, and kde=True adds a kernel density estimate to the plot for a smoother representation of the distribution.
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# KDE
sns.kdeplot(data=tips_df, x='total_bill', fill=True) # fill=True fills the area under the curve for better visualization of the distribution.
```

```python
# KDE- Kernel Density Estimate plot to show the distribution of total bill
sns.kdeplot(data=tips_df, x='total_bill', hue='day', fill=True) # hue='day' colors the density curves based on the day of the week, and fill=True fills the area under the curves for better visualization.
plt.title('KDE Plot of Total Bill by Day of the Week')
plt.xlabel('Total Bill')
plt.ylabel('Density')
plt.show()
```

```python
# pair plot to show relationships between multiple variables

sns.pairplot(tips_df, hue='day') # hue='day' colors the points based on the day of the week, allowing us to see how the relationships between variables differ across days.
plt.suptitle('Pair Plot of Tips Dataset', y=1.02) # Adjust the y position of the title to prevent overlap with the plots
```

```python
# Create a heatmap to show correlation between numerical variables

correlation_matrix = tips_df[['total_bill', 'tip', 'size']].corr() # Calculate the correlation matrix for the numerical variables in the tips dataset.
correlation_matrix
```

```python

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm') # annot=True adds the correlation values to the heatmap, and cmap='coolwarm' specifies the color map to use for the heatmap, where cool colors represent negative correlations and warm colors represent positive correlations.

plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```

```python
### Customization and Styling of Plots

import pandas as pd
sales_data_df = pd.read_csv('100_Sales_Records.csv')
sales_data_df.head()
```

```python
# plot total sales by product

sales_by_product_df = sales_data_df.groupby('Item Type')['Units Sold'].sum().reset_index() # Group the sales data by 'Item Type' and calculate the total 'Units Sold' for each product, then reset the index to create a new DataFrame for plotting.

sales_by_product_df
```

```python
# plot total sales by product using matplotlib

import matplotlib.pyplot as plt

plt.plot(sales_by_product_df['Item Type'], sales_by_product_df['Units Sold'], marker='o', linestyle='-', color='green')

plt.xlabel('Item Type')
plt.ylabel('Units Sold')
plt.title('Total Sales by Product')
plt.xticks(rotation=45) # Rotate the x-axis labels by 45 degrees for better readability, especially if there are many product types or long names.
plt.tight_layout() # Adjust the layout to prevent overlap of labels and ensure everything fits well within the figure.
plt.show()
```

```python
# plot total sales by product using Seaborn

import seaborn as sns

plt.figure(figsize=(10, 6)) # Set the figure size to 10 inches wide and 6 inches tall for better visibility of the plot.

sns.lineplot(data=sales_data_df, x='Item Type', y='Units Sold', estimator=sum) 

# palette='viridis' specifies a color palette for the lines in the line plot, which can enhance the visual appeal and make it easier to distinguish between different product types.
# estimator=sum aggregates the 'Units Sold' values by summing them for each 'Item Type', which is necessary for creating a line plot that shows total sales by product.

plt.xlabel('Item Type')
plt.ylabel('Units Sold')
plt.title('Total Sales by Product')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# barplot to show total sales by product using Seaborn

plt.figure(figsize=(10, 6)) # Set the figure size to 10 inches wide and 6 inches tall for better visibility of the plot.

sns.barplot(data=sales_data_df, x='Item Type', y='Units Sold', estimator=sum) 

# palette='viridis' specifies a color palette for the bars in the bar plot, which can enhance the visual appeal and make it easier to distinguish between different product types.
# estimator=sum aggregates the 'Units Sold' values by summing them for each 'Item Type', which is necessary for creating a bar plot that shows total sales by product.

plt.xlabel('Item Type')
plt.ylabel('Units Sold')
plt.title('Total Sales by Product')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
