```python
### Handling Missing Values
df.isnull()
# but at the very first checking all the cells for missing value is not good practice.
```

```python
#instead first find if any row or colum has any missing cells.
# isnull().any() by default checks by column, if any column has any missing value it will return True otherwise False
df.isnull().any()
```

```python
df.isnull().any(axis=0) # check by column, if any column has any missing value it will return True otherwise False
```

```python
df.isnull().any(axis=1) # check by row
```

```python
# by column how many are missing
df.isnull().sum()
```

```python
# by row how many are missing
df.isnull().sum(axis=1)
```

```python
### fill the missing cells
df_filled = df.fillna(0) # fill the missing cells with 0, and keep a copy of this filled dataframe
df.head(5)
```

```python
df_filled.head(5)
```

```python
df.isnull().sum()
# after filling the missing cells with 0, there are no missing cells in the dataframe, so it will return 0 for all the columns. but i see that there are some missing cells in the 'OrderID' and 'Total Profit'. why ? 

# because the update of the dataframe is not done in place, so the original dataframe is not updated, and the missing cells are still there in the original dataframe. if we want to update the original dataframe then we can use the inplace parameter and set it to True or we can assign the filled dataframe to the original dataframe.
```

```python
df_filled.isnull().sum()
```

```python
### we usually dont put 0 to the missing cells.
# while calculating the mean of the 'Total Profit' column, it will ignore the missing cells and calculate the mean of the available cells, and then we can fill the missing cells with that mean value.
### filling missing cells with the mean of that column.
df['Total Profit_filled'] = df['Total Profit'].fillna(df['Total Profit'].mean())
# a new column 'Total Profit_filled' is created in the dataframe where the missing cells in the 'Total Profit' column are filled with the mean of that column and the rest of the cells are same as the 'Total Profit' column.
```

```python
### Renaming Columns
df_renamed = df.rename(columns={'Order Date':'Ordered Date'}) # it takes a dictionary.
df_renamed.head(6)
```

```python
### renaming rows
df_renamed = df.rename(index={0:'first row', 1:'second row'})
df_renamed.head(6)
```

```python
### change the datatypes -> .astype()
# full meaning of astype() is as a type.
# object datatypes are basically strings
df['Unit Price_int'] = df['Unit Price'].fillna(df['Unit Price'].mean()).astype(int)
```

```python
df.head()
```

```python
### appply function to one of the columns
df['Unit Price_int_2x'] = df['Unit Price_int'].apply(lambda x:x*2)
df.head()
```

```python
### Data Aggregating and Grouping 
# aggregation:
# summarizing data into a single value.
# e.g. sum,mean, of a total column. df['Marks'].mean()

# Grouping:
# groupby() is used to group the data based on some column, and then we can apply some aggregation function on that grouped data. It creates unique values of those column and then apply the aggregation function on that unique values.
# e.g. df.groupby('Dept')
```

```python
# combining this two
# df.groupby('Dept')['Marks'].mean()
grouped_max = df.groupby('Item Type')['Unit Price_int'].max()
# this will group the data by 'Item Type' and then it will calculate the maximum value of 'Unit Price_int' for each group and return a new dataframe with 'Item Type' as index and 'Unit Price_int' as column.
print(grouped_max)
```

```python
### multiple grouping
two_group = df.groupby(['Item Type', 'Country'])['Unit Price_int'].sum()
print(two_group)
```

```python
two_group = df.groupby(['Country','Item Type'])['Unit Price_int'].sum()
print(two_group)
```

```python
whats the difference between these two groupings ?
Think of it like folders

groupby(['Item Type', 'Country'])
Fruit/
    Canada
    USA

Vegetable/
    USA

groupby(['Country', 'Item Type'])
Canada/
    Fruit

USA/
    Fruit
    Vegetable

The underlying records in each group are the same, but the hierarchy (MultiIndex structure) is different.
```

```python
### multiple aggregate func:
mult_aggre = df.groupby('Item Type')['Unit Price_int'].agg(['mean','sum','count'])
mult_aggre
```

```python
### Merging and Joining Dataframes
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, on='Key', how='inner') # inner join will return only the common keys in both dataframes.
```

```python
pd.merge(df1, df2, on='Key', how='outer') # outer join will return all the keys from both dataframes, and if there is no match then it will return NaN.
```

```python
pd.merge(df1, df2, on='Key', how='left') # left join will return all the keys from the left dataframe (df1) and the matching keys from the right dataframe (df2). If there is no match, it will return NaN for the columns from the right dataframe.
```

```python
pd.merge(df1, df2, on='Key', how='right') # right join will return all the keys from the right dataframe (df2) and the matching keys from the left dataframe (df1). If there is no match, it will return NaN for the columns from the left dataframe.
```
