Data manipulation and analysis are key tasks in any data science or data analysis project.
Pandas provides a wide range of functions for data manipulation and analysis, making it easier to
clean, transform and extract insights from data.

```python
### Handling Missing Values
df.isnull()
# but at the very first checking all the cells for missing value is not good.
```

```python
#instead first find if any row or colum has any missing cells.
# isnull() check by column. axis=0
df.isnull().any()
```

```python
# isnull() check by rows. axis=1
df.isnull().any(axis=1)
```

```python
# by column how many are missing
df.isnull().sum()
```

```python
# by row how many are missing
df.isnull().sum(axis=1)
```

```python
### fill the missing cells
df_filled = df.fillna(0)
# and keep a copy of this filled dataframe]
df.head(5)
```

```python
df_filled.head(5)
```

```python
df.isnull().sum()
```

```python
### we usually dont put 0 to the missing cells.
### filling missing cells with the mean of that column.
df['Total Profit_filled'] = df['Total Profit'].fillna(df['Total Profit'].mean())
```

```python
df.head(5)
```

```python
df.dtypes
```

```python
### Renaming Columns
df = df.rename(columns={'Order Date':'Ordered Date'}) #columns takes a dictionary.
df.head()
```

```python
### change the datatypes -> .astype()
# object datatypes are basically strings
df['Unit Price_int'] = df['Unit Price'].fillna(df['Unit Price'].mean()).astype(int)
```

```python
df.head()
```

```python
### appply function to one of the columns
df['Unit Price_int_2x'] = df['Unit Price_int'].apply(lambda x:x*2)
df.head()
```

```python
### Data Aggregating and Grouping
#agreegation: 
# summarizing data into a single value.
# e.g. sum,mean, a total column. df['Marks].mean()

# Grouping:
# dividing data into categories based on some column.
# e.g. df.groupby('Dept')
```

```python
# combining this two
# df.groupby('Dept')['Marks'].mean()
grouped_max = df.groupby('Item Type')['Unit Price_int'].max()
print(grouped_max)
```

```python
### multiple grouping
two_group = df.groupby(['Item Type', 'Country'])['Unit Price_int'].sum()
print(two_group)
```

```python
two_group = df.groupby(['Country','Item Type'])['Unit Price_int'].sum()
print(two_group)
```

```python
### multiple aggregate func:
mult_aggre = df.groupby('Item Type')['Unit Price_int'].agg(['mean','sum','count'])
mult_aggre
```
