### Pandas - Dataframe and Series

Pandas is a powerful data manipulation library in python,
widely used for data analysis and data cleaninng and handling structured data(tables).
It provides two primary data structures:
1) Series -1D (like one -> column)
2) DataFrame -2D (like a full table, multiple -> rows + columns)

A Series is a 
one-dimensional(1D - like a single column in excel/table)
labeled(each value or element has an index) array-like object, 
can hold different data types but internally prefer the uniform data-types.

While a DataFrame is a two-dimensional(2D - rows,columns like excel or full table) 
labeled table (columns has names and rows has indices),
size-mutable(rows, columns can be added or removed) and 
potentially heterogenous(different column can have different data types) 
tabular data stucture with labeled-axes (2 axes -> rows and columns has index and names).


In DataFrame's Each column is a Series
Series -> one axis labeled
DataFrame -> two axes labeled

```python
import pandas as pd
```

### Series (column)

```python
### Series
# a pandas series is a one dimensional array-like object that can hold any datatype but usually uniform datatypes.
# it is similar to a column in a table.
import pandas as pd
data = [1,2,3,4,5]

series = pd.Series(data)
print(series)
print(type(series))


```

```python
## create a series from dictionary
data = {'a':1, 'b':2, 'c':3, 'd':4} # the keys will become the index
series = pd.Series(data)
print(series)
```

```python
# index can be created like this way as well :)
data = [10,20,30]
index = ['abs','bcd','cde']
series = pd.Series(data,index)
print(series)
```

### DataFrames (rows+columns)

```python
### Dataframes
# create a basic dataframe from a dictionary of list
data = {
    'Name': ['krish', 'farabe', 'jhon'],
    'Age': [32,25,45],
    'City': ['Bangalore', 'Mymensingh', 'Florida']
}
dataframe = pd.DataFrame(data)
print(dataframe)
print(type(dataframe))
```

```python
## we can also conver dataframes to 2D array
import numpy as np
print(np.array(dataframe))
```

```python
### create a dataframe from list of dictionary'
data = [
    {'name':'farabe', 'age': 25, 'city': 'Mymensingh'},
    {'name':'krish', 'age': 32, 'city': 'Bangalore'},
    {'name':'jhon', 'age': 45, 'city': 'Florida'}
]

dataframe = pd.DataFrame(data)
print(dataframe)
```
