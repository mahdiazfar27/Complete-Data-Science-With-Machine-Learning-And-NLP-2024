### exp1: temp converter

```python
def convert_temp(temp,unit):
    """Convert temperature between Celsius and Fahrenheit.""" #docstring
    if unit=="c":
        return temp * 9/5 + 32, "Fahrenheit"
    elif unit=="f":
        return (temp-32) * 5/9, "Celsius"
    else:
        return None
```

```python
print(f"{convert_temp(25,'c')[0]} {convert_temp(25,'c')[1]}")
print(f"{convert_temp(77,'f')[0]} {convert_temp(77,'f')[1]}")
```

### exp2 password strength checker

```python
# Write a function that checks if a password is strong. A strong password is defined as one that is at least 8 characters long, contains at least one digit, one lowercase letter, one uppercase letter, and one special character (e.g., !@#$%^&*()_+).

def is_strong_pass(password):
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
    
```

```python
print(is_strong_pass("12345"))
print(is_strong_pass("Strongp@ss26"))
```

### exp3 total cost of a shopping cart

```python
# Write a function that takes a list of dictionaries representing items in a shopping cart and calculates the total cost. Each dictionary contains the keys "name", "unit_price", and "quantity".
def calculate_cost(cart):
    total = 0
    for item in cart:
        total += (item["unit_price"]*item["quantity"])
    return total
```

```python
cart=[
    {"name":"banana", "unit_price": 5, "quantity": 12},
    {"name":"orrange", "unit_price": 15, "quantity": 10},
    {"name":"guava", "unit_price": 8, "quantity": 5},
    {"name":"apple", "unit_price": 5, "quantity": 4},
    {"name":"strawberry", "unit_price": 50, "quantity": 7},
]

print(f"total: {calculate_cost(cart)} tk")
```

### exp4 check if a string is palindrom

```python
# Write a function that checks if a given string is a palindrome. A palindrome is a word, phrase, number, or other sequence of characters that reads the same forward and backward (ignoring spaces, punctuation, and capitalization).
def is_palindrom(s):
    s = s.lower().replace(" ","")
    return s == s[::-1]
```

```python
s = input("write a string")
if(is_palindrom(s)):
    print(s, ": palindrom")
else:
    print(s, ": not palindrom")
```

### exp5 calculate the factorial of a number using recursion

```python
def factorial(num):
    if num == 1:
        return 1
    else:
        return num * factorial(num-1)
```

```python
n = 5
print(factorial(n))
```

### exp6 a func to read a file and count the frequency of each word

```python
# Write a function that takes a file path as input and counts the frequency of each word in the file. The function should return a dictionary where the keys are words and the values are their corresponding frequencies.
def count_word_freq(file_path):
    # i will use a dict to store key & value for the word conunt.
    word_count = {}
    # to open a file we use open func
    with open(file_path,'r') as file: # this will open the file in read mode and assign it to the variable file. the with statement is used to ensure that the file is properly closed after its suite finishes, even if an exception is raised at some point.
        for line in file: # this will iterate through each line in the file. it detect line by the newline character \n.
            words = line.split()
            for word in words:
                word = word.lower().strip('.,!/?;:"\'') # this will convert the word to lowercase and remove any punctuation from the beginning and end of the word. this is done to ensure that words like "Hello" and "hello" are counted as the same word, and that words like "hello," and "hello" are also counted as the same word.
                word_count[word] = word_count.get(word,0)+1
    return word_count
```

```python
file_path = "file_to_read.txt" # as both files are in the same directory
print(count_word_freq(file_path))
```

### exp7 validate email address

```python
# Write a function that validates an email address using regular expressions. The function should return True if the email address is valid and False otherwise.
import re 
# re is a built-in module in Python that provides support for regular expressions. Regular expressions are a powerful tool for matching patterns in strings. They can be used for tasks such as validating input, searching for specific patterns, and replacing substrings. The re module provides functions such as re.match(), re.search(), re.findall(), and re.sub() to work with regular expressions.

def is_email_valid(email):
    pattern = r'^[a-zA-Z0-9._+-]+@[a-zA-Z0-9-]+\.[a-zA-Z0-9-.]+$' 
    # this is a regular expression pattern for validating email addresses. it checks for the following:
    # ^: start of the string
    # [a-zA-Z0-9._+-]+: one or more characters that can be lowercase letters, uppercase letters, digits, dots, underscores, plus signs, or hyphens (the local part of the email)
    # @: the at symbol
    # [a-zA-Z0-9-]+: one or more characters that can be lowercase letters, uppercase letters, digits, or hyphens (the domain name)
    # \.: a literal dot
    # [a-zA-Z0-9-.]+: one or more characters that can be lowercase letters, uppercase letters, digits, dots, or hyphens (the top-level domain)
    # $: end of the string  
    return re.match(pattern,email) is not None
```

```python
print(is_email_valid("mahdiazfar@gmail.com"))
print(is_email_valid("Hello world"))
```
