### map() func

map() is used to apply a func to each item in an iterable(like  list, tuple etc).
its very powerful and heavily used.

```python
### syntax
map(function, iterable)
map(lambda_func, lst)
```

```python
num_lst = [1,2,3,4,5,6]
square_lst = map(lambda x: x**2, num_lst)
print(list(square_lst))
```

map() func returns a map object. we convert it into list

```python
### map multiple iterables
num1 = [1,3,5,7]
num2 = [2,4,6,8]

added_num = list(map(lambda x,y: x+y,num1,num2))  # this will add the corresponding elements of the two lists and return a new list with the results. the first element of the new list will be the sum of the first elements of the two lists, the second element of the new list will be the sum of the second elements of the two lists, and so on. if the two lists have different lengths, the map function will stop when it reaches the end of the shorter list.
print(added_num)
```

```python
### map() to convert list of string to integers
str_num = ['1','2','3','4']
int_num = list(map(int, str_num))
print(int_num)
```

```python
### map() to convert lower words to upper
lower_words = ['hello','world','python','programming']
upper_words = list(map(str.upper, lower_words))
print(upper_words)
```

```python

# map() to extract name from a list of dictionaries
def get_name(person):
    return person['name']

people = [
    {'name': 'krish', 'age': 32},
    {'name': "farabe", 'age': 25},
    {'name': "shammi", 'age': 23}
]
```

```python
names_lst = list(map(get_name,people))
print(names_lst)
```
